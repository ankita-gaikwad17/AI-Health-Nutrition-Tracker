```python
# Import necessary machine learning libraries
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# Sample Dataset: Training data for food items
# Features: [Calories (per 100g), Sugar (g), Fiber (g)]
# Label: 0 = Junk Food, 1 = Healthy Food
data = {
    'Calories': [500, 520, 90, 80, 450, 60, 480, 100],
    'Sugar (g)': [45, 50, 5, 4, 35, 6, 40, 7],
    'Fiber (g)': [1, 2, 6, 7, 2, 5, 1, 6],
    'Label': [0, 0, 1, 1, 0, 1, 0, 1]  # 0: Junk, 1: Healthy
}

# Convert data into a pandas DataFrame
df = pd.DataFrame(data)
X = df[['Calories', 'Sugar (g)', 'Fiber (g)']]
y = df['Label']

# Train the Decision Tree Machine Learning Model
model = DecisionTreeClassifier()
model.fit(X, y)

print("AI Model successfully trained!")

# Test with a new unseen food item:
# Example: Let's test an item with Calories: 95, Sugar: 6g, Fiber: 5g
sample_food = [[95, 6, 5]]
prediction = model.predict(sample_food)

if prediction[0] == 1:
    print("Prediction: This is Healthy Food! ✅")
else:
    print("Prediction: This is Junk Food! ❌")
```

```python
# Test with a junk food item (high calories, high sugar, low fiber)
junk_sample = [[550, 48, 1]]
prediction_2 = model.predict(junk_sample)

if prediction_2[0] == 1:
    print("Prediction: This is Healthy Food! ✅")
else:
    print("Prediction: This is Junk Food! ❌")
```
